```python

import pandas as pd

data = {
    "Student": [
        "Student 1", "Student 2", "Student 3", "Student 4", "Student 5",
        "Student 6", "Student 7", "Student 8", "Student 9", "Student 10"
    ],
    "Mathematics": [14, 12, 16, 10, 15, 13, 17, 11, 14, 18],
    "Physics": [13, 11, 15, 12, 14, 12, 16, 10, 15, 17],
    "Statistics": [15, 13, 17, 11, 16, 14, 18, 12, 14, 16]
}

df = pd.DataFrame(data)

df
```

```python

subjects = ["Mathematics", "Physics", "Statistics"]

print("MEAN")
print(df[subjects].mean())

print("\nMEDIAN")
print(df[subjects].median())

print("\nSTANDARD DEVIATION")
print(df[subjects].std())
```

```python

df["Overall Average"] = df[subjects].mean(axis=1)

df["Status"] = df["Overall Average"].apply(
    lambda x: "Pass" if x >= 10 else "Fail"
)

df[["Student", "Overall Average", "Status"]]
```

```python

pass_rate = (df["Status"] == "Pass").mean() * 100

print(f"Pass rate: {pass_rate:.0f}%")
```

```python

import matplotlib.pyplot as plt

average_scores = df[subjects].mean()

average_scores.plot(kind="bar")

plt.title("Average Performance by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.ylim(0, 20)
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Overall Average"])

plt.title("Overall Average of Each Student")
plt.xlabel("Student")
plt.ylabel("Overall Average")
plt.ylim(0, 20)

plt.xticks(rotation=45)
plt.show()
```

# Statistical Analysis of the Academic Performance of 10 Students

## Introduction

This project presents a statistical analysis of the academic performance of 10 students in Mathematics, Physics, and Statistics.

The project was developed using Python and the Pandas library. Descriptive statistical methods were applied to calculate means, medians, standard deviations, individual averages, and the overall pass rate.

The main purpose of this project is to demonstrate the practical application of Mathematics, Statistics, and Python in the analysis and interpretation of educational data.

## Objectives

The main objectives of this project are:

- To organise academic data from 10 students.
- To calculate descriptive statistical measures.
- To analyse students' individual performance.
- To determine the overall pass rate.
- To create graphical representations of the results.
- To apply Python to a practical data analysis problem.

## Methodology

A quantitative and descriptive approach was used.

The dataset contains the scores of 10 students in three subjects: Mathematics, Physics, and Statistics.

Python was used for data processing and analysis, while Pandas was used to organise the dataset and perform statistical calculations. Matplotlib was used to create graphical representations of the results.

A score of 10 or higher was considered a passing result.

## Results

The analysis produced the following main findings:

- The overall pass rate was 100%.
- All 10 students achieved an overall average of at least 10 points.
- Statistics had the highest average performance among the three subjects.
- The analysis showed differences in individual student performance.
- The graphs provide a visual representation of the academic results.

The descriptive statistics help identify patterns and differences in the students' academic performance.

This project demonstrates how Mathematics, Statistics, and Python can be combined to analyse educational data.

The analysis of 10 students included descriptive statistics, individual averages, classification of academic status, and graphical data visualisation.
